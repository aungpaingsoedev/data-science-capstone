## Load and inspect

Inspect missing values, duplicates, types, and landing outcome distribution.


```python
import pandas as pd
df=pd.read_csv('dataset_part_2.csv')
display(df.head()); display(df.isna().sum())
```

## Clean and label

Fill missing payload mass with the mean, standardize outcomes, and map successful ASDS/RTLS/Ocean outcomes to Class=1.


```python
df['PayloadMass']=df['PayloadMass'].fillna(df['PayloadMass'].mean())
success={'True ASDS','True RTLS','True Ocean'}
if 'Class' not in df: df['Class']=df['Outcome'].isin(success).astype(int)
display(df['Class'].value_counts())
```
