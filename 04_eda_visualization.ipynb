## Prepare

Explore flight experience, payload, site, orbit, and annual landing trends.


```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df=pd.read_csv('dataset_part_2.csv'); df['Date']=pd.to_datetime(df['Date']); df['Year']=df.Date.dt.year
```

## Flight and payload

Color the scatter by Class to reveal the learning effect and payload overlap.


```python
sns.scatterplot(data=df,x='FlightNumber',y='PayloadMass',hue='Class'); plt.show()
```

## Site and orbit

Report sample counts with success rates so small categories are not over-interpreted.


```python
display(df.groupby('LaunchSite')['Class'].agg(['count','mean']).sort_values('mean',ascending=False))
display(df.groupby('Orbit')['Class'].agg(['count','mean']).sort_values('mean',ascending=False))
```

## Time trend

Measure the annual success rate.


```python
df.groupby('Year')['Class'].mean().plot(marker='o',ylabel='Success rate'); plt.show()
```
