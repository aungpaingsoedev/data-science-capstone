## Map sites

Map every launch and color the marker green for success and red for failure. MarkerCluster makes dense sites readable.


```python
import pandas as pd, folium
from folium.plugins import MarkerCluster, MousePosition
df=pd.read_csv('spacex_launch_geo.csv'); m=folium.Map([df.Lat.mean(),df.Long.mean()],zoom_start=5); cluster=MarkerCluster().add_to(m)
for _,r in df.iterrows(): folium.CircleMarker([r.Lat,r.Long],radius=5,color='green' if r['class'] else 'red',fill=True,tooltip=r['Launch Site']).add_to(cluster)
MousePosition().add_to(m); m
```

## Proximity method

Click coastline, railway, highway, and city points in the interactive map; calculate straight-line distance with Haversine and draw labeled PolyLine segments.


```python
from math import radians,sin,cos,asin,sqrt
def haversine(a,b,c,d):
    R=6371; x=radians(c-a); y=radians(d-b); q=sin(x/2)**2+cos(radians(a))*cos(radians(c))*sin(y/2)**2
    return 2*R*asin(sqrt(q))
def add_distance_line(m,start,end,label):
    km=haversine(*start,*end); folium.PolyLine([start,end],tooltip=f'{label}: {km:.2f} km').add_to(m); return km
# Cape sites are coastal and near road infrastructure; VAFB is also coastal. Site selection balances safety and logistics.
```
