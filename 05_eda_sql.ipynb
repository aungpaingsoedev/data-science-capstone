## Create database

Load the cleaned table into SQLite for reproducible SQL analysis.


```python
import pandas as pd, sqlite3
df=pd.read_csv('dataset_part_2.csv'); con=sqlite3.connect(':memory:'); df.to_sql('SPACEXTBL',con,index=False,if_exists='replace')
```

## Queries

Use DISTINCT, WHERE, GROUP BY, AVG, MAX, COUNT, and ORDER BY to answer the course questions.


```python
queries=['SELECT DISTINCT LaunchSite FROM SPACEXTBL','SELECT LaunchSite, COUNT(*) launches, AVG(Class) success_rate FROM SPACEXTBL GROUP BY LaunchSite','SELECT MAX(PayloadMass) max_payload_kg FROM SPACEXTBL','SELECT Orbit, COUNT(*) launches, AVG(Class) success_rate FROM SPACEXTBL GROUP BY Orbit ORDER BY success_rate DESC','SELECT Date, LaunchSite, Outcome FROM SPACEXTBL WHERE Class=1 ORDER BY Date LIMIT 1']
for q in queries: display(pd.read_sql_query(q,con))
```
